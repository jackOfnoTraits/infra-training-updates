```python
import timeit
import numpy as np
```

```python
def benchmark(
    func1,
    func2,
    number=10_000,
    repeat=5,
    name1="Generic Python",
    name2="NumPy",
):
    times1 = timeit.repeat(func1, number=number, repeat=repeat)
    times2 = timeit.repeat(func2, number=number, repeat=repeat)

    best1 = min(times1)
    best2 = min(times2)

    print(f"\n{'Benchmark Results':^45}")
    print("-" * 45)

    for name, best in [(name1, best1), (name2, best2)]:
        print(f"{name}:")
        print(f"  Best run:      {best:.6f} seconds")
        print(f"  Time per call: {best / number:.10f} seconds")

    print("\nComparison:")
    if best1 < best2:
        print(f"{name1} is {best2 / best1:.2f}x faster")
    elif best2 < best1:
        print(f"{name2} is {best1 / best2:.2f}x faster")
    else:
        print("Both have equal measured execution times.")

    return {name1: times1, name2: times2}
```

```python
# One-dimensional array

arr = [10, 20, 30, 40, 50]
np_arr = np.array(arr)

def python_1d():
    [x + 10 for x in arr]

def numpy_1d():
    np_arr + 10

python_1d(); numpy_1d()

print("Generic Python result:", [x + 10 for x in arr])
print("NumPy result:", np_arr + 10)

benchmark(python_1d, numpy_1d, name1="Python 1D", name2="NumPy 1D")
```

```python
# Two-dimensional array

arr_2d = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
np_arr_2d = np.array(arr_2d)

def python_2d():
    [[x + 10 for x in row] for row in arr_2d]

def numpy_2d():
    np_arr_2d + 10

print("Generic Python result:", [[x + 10 for x in row] for row in arr_2d])
print("NumPy result:\n", np_arr_2d + 10)

benchmark(python_2d, numpy_2d, name1="Python 2D", name2="NumPy 2D")
```

```python
# Three-dimensional array

arr_3d = [[[11, 12, 7], [9, 5, 6]], [[7, 8, 9], [10, 4, 3]]]
np_arr_3d = np.array(arr_3d)

def python_3d():
    [[[x + 10 for x in row] for row in block] for block in arr_3d]

def numpy_3d():
    np_arr_3d + 10

print("Generic Python result:", [[[x + 10 for x in row] for row in block] for block in arr_3d])
print("NumPy result:\n", np_arr_3d + 10)

benchmark(python_3d, numpy_3d, name1="Python 3D", name2="NumPy 3D")
```

```python
# Computing sum

def python_3d_sum():
    sum([num for matrix in arr_3d for row in matrix for num in row])

def numpy_3d_sum():
    np_arr_3d.sum()

print("Generic Python result:", sum([num for matrix in arr_3d for row in matrix for num in row]))
print("NumPy result:\n", np_arr_3d.sum())

benchmark(python_3d_sum, numpy_3d_sum, name1="Python 3D", name2="NumPy 3D")
```

```python
# Finding min

def python_3d_min():
    min([num for matrix in arr_3d for row in matrix for num in row])

def numpy_3d_min():
    np_arr_3d.min()

print("Generic Python result:", min([num for matrix in arr_3d for row in matrix for num in row]))
print("NumPy result:\n", np_arr_3d.min())

benchmark(python_3d_min, numpy_3d_min, name1="Python 3D", name2="NumPy 3D")
```

```python
# Finding max

def python_3d_max():
    max([num for matrix in arr_3d for row in matrix for num in row])

def numpy_3d_max():
    np_arr_3d.max()

print("Generic Python result:", max([num for matrix in arr_3d for row in matrix for num in row]))
print("NumPy result:\n", np_arr_3d.max())

benchmark(python_3d_max, numpy_3d_max, name1="Python 3D", name2="NumPy 3D")
```

```python
# Finding mean

def python_3d_mean():
    numbers = [
        num 
        for matrix in arr_3d 
        for row in matrix 
        for num in row
    ]
    sum(numbers) / len(numbers)

def numpy_3d_mean():
    np_arr_3d.mean()

numbers = [num for matrix in arr_3d for row in matrix for num in row]
print("Generic Python result:", sum(numbers) / len(numbers))
print("NumPy result:\n", np_arr_3d.mean())

benchmark(python_3d_mean, numpy_3d_mean, name1="Python 3D", name2="NumPy 3D")
```
